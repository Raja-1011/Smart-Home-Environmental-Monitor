```python
import re
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Path to your file
file_path = Path("/Users/rajip/Downloads/Sensor reading(TI).txt")

# Read file
text = file_path.read_text(errors="ignore")

# Extract values from log
brightness = [float(x) for x in re.findall(r"Brightness:\s*([\d.]+)\s*%", text)]
gas        = [float(x) for x in re.findall(r"Gas Level:\s*([\d.]+)\s*%", text)]
temp       = [float(x) for x in re.findall(r"TEMP:\s*([\d.]+)\s*C", text)]
humidity   = [float(x) for x in re.findall(r"HUM:\s*([\d.]+)\s*%RH", text)]

# Create time axis (sample index)
samples = np.arange(len(brightness))

# Plot all sensors on one graph
plt.figure(figsize=(10, 6))

plt.plot(samples, brightness, label="Brightness (%)")
plt.plot(samples, gas, label="Gas level (%)")
plt.plot(samples, temp, label="Temperature (°C)")
plt.plot(samples, humidity, label="Humidity (%RH)")

plt.xlabel("Sample")
plt.ylabel("Sensor value")
plt.title("Combined sensor readings over time")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
