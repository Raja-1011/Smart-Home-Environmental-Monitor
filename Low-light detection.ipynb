```python
import re
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Path to your file
file_path = Path("/Users/rajip/Downloads/low-light.txt")

# Read file
text = file_path.read_text(errors="ignore")

# Extract Brightness values from the log
pattern = re.compile(r"Brightness:\s*([\d.]+)\s*%")
brightness = [float(x) for x in pattern.findall(text)]

# Create sample index
samples = np.arange(len(brightness))

# Threshold
threshold = 10.0

# Plot
plt.figure(figsize=(8, 5))
plt.plot(samples, brightness)
plt.axhline(threshold, linestyle="--")

plt.xlabel("Sample")
plt.ylabel("Brightness (%)")
plt.title("Low-light detection (threshold=10%) — low-light.txt")

plt.tight_layout()
plt.show()
```
